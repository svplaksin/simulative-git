### **Решения**

#### **Задача 1**

Ваша задача - выяснить сколько в среднем тратится времени на решение задачи.

**Примечание**: для правильного подсчета - рассчитайте сначала среднее время решения по каждой задаче в отдельности, и только затем находите общее среднее время решения задач.

Результат - число типа `float`, округлите до 2 знаков после запятой и запишите в переменную `res`.


**Решение**

Напишите свое решение ниже

```python
# Ваше решение здесь
# Преобразуем столбец из объекта в DateTime
df['created_at'] = pd.to_datetime(df['created_at'])

# Чтобы найти время первой попытки юзера решить задачу, группируем их, и выбираем минимальььное время
first_attempt = df.groupby(['user_id', 'problem_id'])['created_at'].min()

# Чтобы найти время первой правильной попытки, сначала фильтруем только все правильные, и их них берем минимальное время
first_correct = df[df['is_correct'] == 1.0].groupby(['user_id', 'problem_id'])['created_at'].min()

# Находим время решения задачи. Если задача не решена юзером - дропаем ее
solution_time = (first_correct - first_attempt).dropna()

# убираем решенные с первой попытки записи, то есть те, у которых разность времени между первым правильным решением и первой попыткой равна 0
solution_time = solution_time[solution_time > pd.Timedelta(0)]

# Ищем сначала среднее время решения по задаче
avg_by_problem = solution_time.groupby(level='problem_id').mean()

# Считаем общее среднее по юзерам
overall_avg = avg_by_problem.mean()

# Приводим результат к секундам, с точность до 2 знаков
res = round(overall_avg.total_seconds(), 2)
```

#### **Задача 2**

Ваша задача - выяснить сколько часов в среднем проводит юзер в день на платформе. Перерывы в активности за день - не учитываем.

Результат - число типа `float`, округлите до 2 знаков после запятой и запишите в переменную `res2`.

**Решение**

Напишите свое решение ниже

```python
# Ваше решение здесь

# Добавляем столбец с датами
df['date'] = df['created_at'].dt.date

# Сортируем сначала по юзерам, потом по дате, и внутри дат по времени
df = df.sort_values(['user_id','date','created_at'])
```

```python
df
```

```python
# Чтобы найти время активности юзера, вычитаем из последней активности первую
last_user_activity = df.groupby(['user_id', 'date'])['created_at'].max()
first_user_activity = df.groupby(['user_id', 'date'])['created_at'].min()
total_user_activity = last_user_activity - first_user_activity

# Считаем среднее
avg_activity = total_user_activity.mean()

# Округляем
res2 = round(avg_activity.total_seconds() / 3600, 2)
```

#### **Задача 3**

Теперь давайте посмотрим на активные сеансы. Выясните, сколько задач в среднем решается за один активный сеанс.

**Активный сеанс** - период, когда между любой активностью пользователя разница менее или равна часу, не более

**Важно**: в расчет берем не только успешные попытки решений (`is_correct=1`), а и неуспешные тоже (`is_correct=0`), и тип `run` в том числе.

Результат - число типа `float`, округлите до 2 знаков после запятой и запишите в переменную `res3`.

**Решение**

Напишите свое решение ниже

```python
# Ваше решение здесь

# Для расчета интервалов создадим столбец со сдвинутыми на 1 строку вниз временами с помощью shift(1). Но сдвигать нужно отдельно по юзерам и в рамках 1 дня
# NaT - если предыдущего события нет
df['previous_time'] = df.groupby(['user_id', 'date'])['created_at'].shift(1)

# Считаем интервалы между событиями юзеров в рамках 1 дня
df['time_diff'] = df['created_at'] - df['previous_time']

# Создаём маркер начала сеанса
df['new_session'] = (df['time_diff'].isna()) | (df['time_diff'] > pd.Timedelta(hours=1))

# Присваиваем сеансам номера с помощью cumsum()
df['session_id'] = df.groupby('user_id')['new_session'].cumsum()

# Считаем количество задач за сеанс
problems_per_session = df.groupby(['user_id', 'session_id'])['problem_id'].nunique()

# Вычисляем итоговое среднее
res3 = round(problems_per_session.mean(), 2)
```

#### **Задача 4**

И финальная - найдите самый "популярный" час дня на нашей платформе.

Популярность определяем максимальным количеством уникальных пользователей, совершающих какую-либо активность в этот период

Результат в числовом формате запишите в переменную `res4`.

Например, самым популярным часом стал период с 22 до 23, тогда в переменной `res4` должно лежать **22**. Обозначающее начало этого периода.

**Решение**

Напишите свое решение ниже

```python
# Ваше решение здесь

# Выделяем часы из временной метки
df['hour'] = df['created_at'].dt.hour

# Считаем уникальных пользователей в каждом часе
users_per_hour = df.groupby('hour')['user_id'].nunique()

# Выбираем самый популярный час
res4 = users_per_hour.idxmax()
```
